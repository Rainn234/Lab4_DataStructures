```python
from collections import defaultdict

# Setup Data Generation Inputs
LAST_NAME = "Barquin"
FAVORITE_ARTIST = "Bruno Mars"
SEED_NUM = sum(ord(c) for c in FAVORITE_ARTIST)

student_labels = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# Performance Tier Categorization
def get_category(score):
    if score is None:
        return "INCOMPLETE"
    if score >= 90:
        return "EXCELLENT"
    if score >= 75:
        return "SATISFACTORY"
    return "NEEDS IMPROVEMENT"

# 1. Generate Performance Records
raw_records = []
char_positions = [ord(c) for c in FAVORITE_ARTIST]

idx = 0
for student in student_labels:
    for subject in subjects:
        score = (SEED_NUM + char_positions[idx % len(char_positions)]) % 51 + 50
        raw_records.append({"student": student, "subject": subject, "score": score})
        idx += 1

# Inject Repeated & Incomplete Edge Cases
raw_records.append(raw_records[0].copy())  # Repeated record
raw_records.append({"student": student_labels[1], "subject": "PHYSICS", "score": None})  # Incomplete record

# 2. De-duplication & Error Detection
seen = set()
unique_records = []
repeated_records = []
incomplete_records = []

for rec in raw_records:
    rec_tuple = (rec["student"], rec["subject"], rec["score"])
    if rec["score"] is None:
        incomplete_records.append(rec)
    if rec_tuple in seen:
        repeated_records.append(rec)
    else:
        seen.add(rec_tuple)
        unique_records.append(rec)

# 3. Summaries & Analysis
student_scores = defaultdict(list)
subject_scores = defaultdict(list)

for rec in unique_records:
    if rec["score"] is not None:
        rec["category"] = get_category(rec["score"])
        student_scores[rec["student"]].append(rec["score"])
        subject_scores[rec["subject"]].append(rec["score"])

# 4. Report Presentation
print("=== ADDITIONAL EXERCISE 2: STUDENT PERFORMANCE REPORT ===")
print(f"Total Raw Records Processed: {len(raw_records)}")
print(f"Repeated Records Found: {len(repeated_records)}")
print(f"Incomplete Records Found: {len(incomplete_records)}\n")

print("--- STUDENT AVERAGES ---")
for student, scores in student_scores.items():
    avg = sum(scores) / len(scores) if scores else 0
    print(f"{student}: Average = {avg:.2f}")

print("\n--- SUBJECT AVERAGES ---")
for subject, scores in subject_scores.items():
    avg = sum(scores) / len(scores) if scores else 0
    print(f"{subject}: Average = {avg:.2f}")
```

```python
from collections import defaultdict

LAST_NAME = "Barquin"
FAVORITE_ARTIST = "Bruno Mars"
SEED_NUM = sum(ord(c) for c in FAVORITE_ARTIST)

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 20, "USB": 300, "HEADPHONES": 800, "CALCULATOR": 500}

# 1. Generate Transactions
transactions = []
char_positions = [ord(c) for c in FAVORITE_ARTIST]

for i in range(8):
    cust = customers[i % len(customers)]
    prod = products[i % len(products)]
    qty = (SEED_NUM + char_positions[i % len(char_positions)]) % 5 + 1
    transactions.append({"customer": cust, "product": prod, "qty": qty})

# Add at least one duplicate customer-product purchase
transactions.append({"customer": customers[0], "product": products[0], "qty": 2})

# 2. Process Values and Structure Summaries
customer_totals = defaultdict(float)
product_totals = defaultdict(float)
seen_transactions = set()
repeated_purchases = []

total_revenue = 0.0

for txn in transactions:
    key = (txn["customer"], txn["product"])
    if key in seen_transactions:
        repeated_purchases.append(txn)
    seen_transactions.add(key)
    
    val = txn["qty"] * prices[txn["product"]]
    txn["total_value"] = val
    
    customer_totals[txn["customer"]] += val
    product_totals[txn["product"]] += val
    total_revenue += val

# 3. Consolidated Output
print("=== ADDITIONAL EXERCISE 3: CONSOLIDATED TRANSACTION REPORT ===")
print(f"Total Transactions Processed: {len(transactions)}")
print(f"Repeated Purchases Detected: {len(repeated_purchases)}")
print(f"Overall Store Revenue: ₱{total_revenue:,.2f}\n")

print("--- CUSTOMER SPENDING SUMMARY ---")
for cust, amt in sorted(customer_totals.items(), key=lambda x: x[1], reverse=True):
    print(f"{cust}: ₱{amt:,.2f}")

print("\n--- PRODUCT SALES SUMMARY ---")
for prod, amt in sorted(product_totals.items(), key=lambda x: x[1], reverse=True):
    print(f"{prod}: ₱{amt:,.2f}")
```
